```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    input_variables=["input"],
    template=(
        "You are an expert AI engineer. "
        "Provide the best possible answer.\n\n"
        "Question:\n{input}"
    )
)
prompt
```

```python
## using this prompt with the llm model 
chain = prompt | llm
response = chain.invoke({"input" : "can u tell me about langsmith"})
print(response)
```

```python
## string output parser

from langchain_core.output_parsers import StrOutputParser
output_parser = StrOutputParser()
chain = prompt|llm|output_parser

response = chain.invoke({"input":"can you tell me about langsmith"})
print(response)
```

```python
import os
from dotenv import load_dotenv
load_dotenv()


os.environ['LANGCHAIN_API_KEY'] = os.getenv("LANGCHAIN_API_KEY")
os.environ['GEMINI_API_KEY'] = os.getenv("GEMINI_API_KEY")
os.environ['LANGCHAIN_PROJECT']  = os.getenv("LANGCHAIN_PROJECT")

os.getenv("GEMINI_API_KEY")
```

```python
##IMPORTED THE GOOGLE GEMINI MODEL

from langchain_google_genai import ChatGoogleGenerativeAI

geminiLLM = ChatGoogleGenerativeAI(
   model="gemini-2.5-flash",
    api_key=os.environ["GEMINI_API_KEY"]
)

geminiLLM
```

```python
from langchain_core.prompts import ChatPromptTemplate
 
prompt = ChatPromptTemplate.from_messages([
    ("system" , "You are modi , the Prime Minister of India"),
    ("user" , "{input}")
])


## making the chain which connects prompt with llm
chain = prompt | geminiLLM


## running the chain 

response = chain.invoke({"input" : "explain the rag architecture in 1 sentence"})
print(response)
```
