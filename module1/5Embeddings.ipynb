```python
from langchain_community.document_loaders import TextLoader
## first we load the text using text loader
from langchain_text_splitters import RecursiveCharacterTextSplitter

## recursive character text splitter is used to split the text into smaller chunks 

text_loader = TextLoader("speech.txt")
## Text Loader is a class from Langchain that opens a text file and turns it into a document object that Langchain can undertand 

## it basically opens the file 
## it reads all the text inside 
## it wraps it into a special object called document 


docs = text_loader.load()
## then using .load
## this actually performs the action of opening the file 
## reading its contents 
## storing the contents inside a document object 
## returns list of documents(usually a list with 1 document for text files )


text_splitter = RecursiveCharacterTextSplitter(chunk_size = 500 , chunk_overlap = 50)
## we split the text into chunks
## each chunk will have 500 characters 
## chunk will overlap by 50 characters to preserve continuity 

final_docs = text_splitter.split_documents(docs)
## this takes the entire speech and breaks it into smaller pieces and it is useful for embeddings , vector stores , retrieval etc


len(final_docs)
final_docs


texts  = []
for doc in final_docs : 
    texts.append(doc.page_content)

texts
final_docs
## shorter and cleaner way would be 
#texts = [doc.page_content for doc in final_docs]
```

```python

from google.genai import Client
## Client is a class from google.genai 
## this is like a template that defines how to connect to gemini 
## what methods are available ; how reqs and responses work ! 

client = Client(api_key=os.environ["GEMINI_API_KEY"])
## this line help us creating an instance of a client class 
## this creates an object 
## when we pass over the keys , now it becomes a fully usable object that can send requests to the Gemini's servers 
## this is a usable object now 
## it stores all the configuration like api key , server endpoints , settings , available models 

result = client.models.embed_content(
        model="gemini-embedding-001",
        contents=texts)
embeddings = result.embeddings
## inside the client object we have different submodules or interfaces - one of them is client.models 
## it contains stuff like embed_content , generate_content , count_token 
## for contensts , we send the text chunks that is texxts 
## so it works like 
# ->>>> i send data (texts) with model name , api key and text chunk 
# ->>> client prepares a https request to google's servers 
# -->>> google's embedding model preocess the text and converts it into high dimensional numerical vectors 
# ->>>> server returns the embeddings (reponse is a structured json like object) 
# ->>>> sdk converts json into a python object and that gets stored in the result 

print(type(embeddings))
print(len(embeddings))
print(type(embeddings[0]))
##print(len(embeddings[0]))


processed_embeddings = [e.values for e in embeddings]
processed_embeddings
```

```python
##import chromadb 

import chromadb
from chromadb.config import Settings 

## create/connect to chroma client 
chroma_client = chromadb.Client()

## this is used to create a collection
collection = chroma_client.get_or_create_collection(name="my_collection_list")


## now we prepare the ids and metdata that will be used inside chromadb 


ids = [f"chunk-{i}" for i in range(len(texts))]
metadatas = [doc.metadata for doc in final_docs]

collection.add(
    ids=ids,
    embeddings=processed_embeddings,
    documents=texts,
    metadatas=metadatas
)



query="where was modi born ? "


result = client.models.embed_content(
        model="gemini-embedding-001",
        contents=[query])




result_query  = result.embeddings[0].values
result_query


queryFromChroma = collection.query(
    query_embeddings=result_query,
    n_results=3
)

queryFromChroma


```
